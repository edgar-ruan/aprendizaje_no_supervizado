```python
# No mostrar advertencias
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./data/retailmax.csv')
df.head()
```

### 1. ¿Cuál es la distribución de edades de los clientes?
- **Objetivo**: Comprender la distribución de las edades en el conjunto de datos.
- **Respuesta esperada**: Un histograma de la columna Age.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], bins=15, kde=True)
plt.title('Distribución de edades')
plt.xlabel('Edad')
plt.ylabel('Número de clientes')
plt.show()
```

**Respuesta:** La mayoría de los clientes tienen entre 20 y 50 años. Hay más clientes jóvenes que mayores de 60.

### 2. ¿Existen diferencias significativas en los ingresos anuales entre hombres y mujeres?
- **Objetivo**: Comparar la distribución de ingresos anuales entre géneros.
- **Respuesta esperada**: Una gráfica de cajas (boxplot) que muestre la distribución de Annual Income (k$) por género.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Gender', y='Annual Income (k$)')
plt.title('Ingreso anual por género')
plt.show()

df.groupby('Gender')['Annual Income (k$)'].describe()
```

**Respuesta:** Los ingresos de hombres y mujeres son muy parecidos. No se ve una diferencia significativa.

### 3. ¿Cómo se distribuye la puntuación de gasto (Spending Score) entre los diferentes rangos de edad?
- **Objetivo**: Analizar la relación entre la edad y la puntuación de gasto.
- **Respuesta esperada**: Un scatter plot o boxplot que compare la Spending Score (1-100) con diferentes grupos de edad.

```python
df['Age Group'] = pd.cut(df['Age'], bins=[17, 25, 35, 45, 55, 70],
                         labels=['18-25', '26-35', '36-45', '46-55', '56-70'])

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Age Group', y='Spending Score (1-100)')
plt.title('Spending Score por grupo de edad')
plt.show()
```

**Respuesta:** Los clientes jóvenes (18-35) tienden a tener puntuaciones de gasto más altas. Después de los 45 baja bastante.

### 4. ¿Cuál es la correlación entre el ingreso anual y la puntuación de gasto?
- **Objetivo**: Identificar si existe una relación lineal entre el ingreso y el gasto.
- **Respuesta esperada**: Un scatter plot y el coeficiente de correlación.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)')
plt.title('Ingreso anual vs Spending Score')
plt.show()

corr = df['Annual Income (k$)'].corr(df['Spending Score (1-100)'])
print(f'Coeficiente de correlación: {corr:.4f}')
```

**Respuesta:** La correlación es casi 0, entonces no hay relación lineal. Pero en la gráfica se ven como 5 "nubes" de puntos, lo que sugiere que sí hay grupos.

### 5. ¿Cómo varía la puntuación de gasto en diferentes grupos de ingresos anuales?
- **Objetivo**: Examinar cómo se comportan en gasto los clientes de distintos rangos de ingreso.
- **Respuesta esperada**: Un boxplot o violín de Spending Score por rangos de Annual Income (k$).

```python
df['Income Group'] = pd.cut(df['Annual Income (k$)'], bins=[0, 40, 70, 100, 140],
                            labels=['Bajo (<40k)', 'Medio (40-70k)', 'Alto (70-100k)', 'Muy alto (>100k)'])

plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Income Group', y='Spending Score (1-100)')
plt.title('Spending Score por grupo de ingreso')
plt.show()
```

**Respuesta:** En ingreso medio el gasto se concentra cerca de 50. En ingresos bajos y altos hay de todo: gente que gasta mucho y gente que gasta poco.

### 6. ¿Cuál es la proporción de clientes por género?
- **Objetivo**: Determinar el balance de género en el conjunto de datos.
- **Respuesta esperada**: Una gráfica de barras o de pastel.

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6, 6))
plt.title('Proporción de clientes por género')
plt.ylabel('')
plt.show()
```

**Respuesta:** Hay un poco más de mujeres (56%) que hombres (44%).

### 7. ¿Qué grupos de edad gastan más en promedio?
- **Objetivo**: Identificar los grupos de edad con mayor puntuación de gasto promedio.
- **Respuesta esperada**: Una gráfica de barras del gasto promedio por grupo de edad.

```python
gasto_edad = df.groupby('Age Group')['Spending Score (1-100)'].mean()

gasto_edad.plot(kind='bar', figsize=(8, 5))
plt.title('Spending Score promedio por grupo de edad')
plt.ylabel('Spending Score promedio')
plt.show()

gasto_edad
```

**Respuesta:** Los grupos 18-25 y 26-35 son los que más gastan en promedio.

### 8. ¿Hay alguna relación entre la edad y el ingreso anual de los clientes?
- **Objetivo**: Explorar si hay una tendencia entre edad e ingreso.
- **Respuesta esperada**: Un scatter plot de Age vs Annual Income (k$).

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Annual Income (k$)')
plt.title('Edad vs Ingreso anual')
plt.show()

print(f"Correlación: {df['Age'].corr(df['Annual Income (k$)']):.4f}")
```

**Respuesta:** No hay una tendencia clara. Los ingresos más altos están más o menos entre los 30 y 50 años.

### 9. ¿Cuál es la distribución conjunta de la edad y el ingreso anual?
- **Objetivo**: Entender cómo se distribuyen estas dos variables en conjunto.
- **Respuesta esperada**: Un gráfico de hexágonos de Age y Annual Income (k$).

```python
sns.jointplot(data=df, x='Age', y='Annual Income (k$)', kind='hex', height=7)
plt.show()
```

**Respuesta:** La zona más densa está entre 20 y 50 años con ingresos de 40k a 80k.

### 10. ¿Cómo se distribuyen los clientes en función de la puntuación de gasto y el género?
- **Objetivo**: Analizar la relación entre la puntuación de gasto y el género.
- **Respuesta esperada**: Un violín de Spending Score por género.

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Gender', y='Spending Score (1-100)')
plt.title('Spending Score por género')
plt.show()
```

**Respuesta:** Las distribuciones son muy parecidas. El género por sí solo no explica cuánto gasta un cliente.
